# Python ile Dosya Yönetimi ve Veri Kalıcılığı (File I/O)

```python
# Eğer kutuphane.txt dosyası yoksa, test için örnek kitaplarla otomatik oluşturur
ornek_veri = """Kürk Mantolu Madonna;Sabahattin Ali
İçimizdeki Şeytan;Sabahattin Ali
Tutunamayanlar;Oğuz Atay
Çalıkuşu;Reşat Nuri Güntekin"""

import os

# Dosya yoksa oluşturup içine örnek veriyi yazar
if not os.path.exists("kutuphane.txt"):
    with open("kutuphane.txt", "w", encoding="utf-8") as f:
        f.write(ornek_veri)
    print("kutuphane.txt dosyası örnek verilerle oluşturuldu.")
else:
    print("kutuphane.txt dosyası zaten mevcut.")
```

```python
kitaplar = []

# Dosyadan mevcut kitapları okuma ve liste haline getirme
with open("kutuphane.txt", "r", encoding="utf-8") as f:
    for satir in f:
        satir = satir.strip()
        if satir:
            parcalar = satir.split(";")
            kitap_adi = parcalar[0]
            yazar_adi = parcalar[1]
            kitaplar.append([kitap_adi, yazar_adi])

while True:
    print("\n1 - Kitap Ekle")
    print("2 - Kitap Çıkar")
    print("3 - Kitapları Listele")
    print("4 - Çıkış")

    secim = input("Lütfen yapmak istediğiniz işlemi seçin (1-4): ")

    if secim == "1":
        kitap_adi = input("Kitap Adı: ").strip()
        yazar_adi = input("Yazar Adı: ").strip()

        var_mi = False
        for k in kitaplar:
            if k[0].lower() == kitap_adi.lower() and k[1].lower() == yazar_adi.lower():
                var_mi = True
                break

        if var_mi:
            print("Bu kitap zaten kütüphanede kayıtlı.")
        else:
            kitaplar.append([kitap_adi, yazar_adi])

            with open("kutuphane.txt", "a", encoding="utf-8") as f:
                f.write(f"{kitap_adi};{yazar_adi}\n")

            print(f"{kitap_adi} kitabı kütüphaneye eklendi.")

    elif secim == "2":
        yazar_arama = input("Almak istediğiniz kitabın yazarının adını girin: ").strip()

        yazarin_kitaplari = []
        for k in kitaplar:
            if k[1].lower() == yazar_arama.lower():
                yazarin_kitaplari.append(k)

        if len(yazarin_kitaplari) == 0:
            print("Bu yazara ait kitap bulunamadı.")
        else:
            sira = 1
            for k in yazarin_kitaplari:
                print(f"{sira} - {k[0]}")
                sira += 1

            secilen_numara = input("Almak istediğiniz kitabın numarasını girin: ")

            if secilen_numara.isdigit():
                secilen_numara = int(secilen_numara)

                if 1 <= secilen_numara <= len(yazarin_kitaplari):
                    silinecek_kitap = yazarin_kitaplari[secilen_numara - 1]

                    kitaplar.remove(silinecek_kitap)

                    with open("kutuphane.txt", "w", encoding="utf-8") as f:
                        for k in kitaplar:
                            f.write(f"{k[0]};{k[1]}\n")

                    print(f"{silinecek_kitap[0]} adlı kitabı aldınız. İyi okumalar.")
                else:
                    print("Geçersiz seçim.")
            else:
                print("Geçersiz seçim.")

    elif secim == "3":
        if len(kitaplar) == 0:
            print("Kütüphanede kayıtlı kitap bulunmamaktadır.")
        else:
            sira = 1
            for k in kitaplar:
                print(f"{sira} - {k[0]} - {k[1]}")
                sira += 1

    elif secim == "4":
        break

    else:
        print("Geçersiz seçim.")
```
