# Sayı Tahmin Algoritması ve Koşullu İfadeler

```python
X = 90

print("Aklımdan bir sayı tuttum, hadi bil!")
Y = int(input("Bir sayı giriniz: "))

sonuc = (Y * Y) - Y

if sonuc == X:
    print("100 puan kazandınız.")
else:
    fark = sonuc - X
    if fark < 0:
        fark = -fark
    if fark <= 10:
        print("50 puan kazandınız.")
    elif fark <= 50:
        print("Yaklaştın, yeniden dene.")
    else:
        print("Kaybettiniz.")
```
