# Temel Veri Yapıları ve Kütüphane Yönetim Konsol Uygulaması

```python
kutuphane = [
    ["Kürk Mantolu Madonna", "Sabahattin Ali"],
    ["İçimizdeki Şeytan", "Sabahattin Ali"],
    ["Tutunamayanlar", "Oğuz Atay"],
    ["Çalıkuşu", "Reşat Nuri Güntekin"]
]

while True:
    print("\n--- KÜTÜPHANE UYGULAMASI ---")
    print("1 - Kitap Ekle")
    print("2 - Kitap Çıkar")
    print("3 - Kitapları Listele")
    print("4 - Çıkış")

    secim = input("Lütfen bir seçim yapın (1-4): ")

    if secim == "1":
        kitap_adi = input("Eklenecek kitabın adı: ")
        yazar_adi = input("Yazarın adı: ")

        var_mi = False

        for kitap in kutuphane:
            if kitap[0].lower() == kitap_adi.lower():
                if kitap[1].lower() == yazar_adi.lower():
                    var_mi = True
                    break

        if var_mi == True:
            print("Bu kitap zaten kütüphanede kayıtlı.")
        else:
            kutuphane.append([kitap_adi, yazar_adi])
            print(kitap_adi + " kitabı kütüphaneye eklendi.")

    elif secim == "2":
        yazar_ara = input("Almak istediğiniz kitabın yazarının adını girin: ")

        yazarin_kitaplari = []

        for kitap in kutuphane:
            if kitap[1].lower() == yazar_ara.lower():
                yazarin_kitaplari.append(kitap)

        if len(yazarin_kitaplari) == 0:
            print("Bu yazara ait kitap bulunamadı.")
        else:
            print("\n" + yazar_ara + " yazarına ait kitaplar:")
            sira = 1
            for kitap in yazarin_kitaplari:
                print(str(sira) + " - " + kitap[0])
                sira = sira + 1

            secilen_no = input("Almak istediğiniz kitabın numarasını girin: ")
            secilen_index = int(secilen_no) - 1

            if 0 <= secilen_index < len(yazarin_kitaplari):
                silinecek_kitap = yazarin_kitaplari[secilen_index]
                kutuphane.remove(silinecek_kitap)
                print(silinecek_kitap[0] + " adlı kitabı aldınız. İyi okumalar.")
            else:
                print("Geçersiz seçim.")

    elif secim == "3":
        if len(kutuphane) == 0:
            print("Kütüphanede kayıtlı kitap bulunmamaktadır.")
        else:
            print("\n--- KÜTÜPHANEDEKİ KİTAPLAR ---")
            sira = 1
            for kitap in kutuphane:
                print(str(sira) + " - " + kitap[0] + " - " + kitap[1])
                sira = sira + 1

    elif secim == "4":
        print("Çıkış yapılıyor...")
        break

    else:
        print("Geçersiz seçim, tekrar deneyin.")
```
