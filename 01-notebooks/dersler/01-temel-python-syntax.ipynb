# Hesap Makinesi

```python
sayi_1 = float(input("Sayı : "))
sembol = input("Yapılacak işlem : ")
sayi_2 = float(input("Sayı : "))

if sembol == "+":
  print(sayi_1+sayi_2)
elif sembol == "-":
  print(sayi_1-sayi_2)
elif sembol == "*":
  print(sayi_1*sayi_2)
elif sembol == "/":
  print(sayi_1/sayi_2)
else:
  print("Hatalı İşlem!")
```

# Hesaba Giriş Uygulaması


```python
kullanici_adi = input("Kullanıcı Adı : ")
sifre = input("Şifre : ")

belirlenmis_kullanici_adi = "ecrinky"
belirlenmis_sifre = "Python"

if belirlenmis_kullanici_adi != kullanici_adi or belirlenmis_sifre != sifre:
  print("Kullanıcı adınız veya şifreniz hatalı!")
else:
  print("Giriş başarılı!")
```

# Belge Hesaplama Aracı

```python
ögrencinin_notu = float(input("Lütfen notunuzu giriniz : "))

if ögrencinin_notu >= 85 and ögrencinin_notu <= 100:
  print("Takdir belgesi almaya hak kazandınız.")
elif ögrencinin_notu >= 70 and ögrencinin_notu < 85:
  print("Teşekkür belgesi almaya hak kazandınız.")
elif ögrencinin_notu < 70 and ögrencinin_notu >= 45:
  print("Sınıfı belgesiz geçtiniz.")
elif ögrencinin_notu < 45 and ögrencinin_notu >= 0:
  print("Sınıfta kaldınız.")
else:
  print("Hatalı Değer!")
```
