```python
def hipo(a, b, c):

  if a**2 + b**2 == c**2:
    return "Bu üçgen dik üçgendir."
  else:
    return "Bu üçgen dik değildir."

a = int(input("a değerini girin : "))
b = int(input("b değerini girin : "))
c = int(input("c değerini girin : "))
hipo(a, b, c)
```

```python
rehber = {
    "ecrin":{
        "cep":5555555555,
        "is":1212121212,
        "ev":887788778877
    },
    "ahmet":{
        "cep":4444444444,
        "is":12515151515,
        "ev":997799779977
    },
}

while True:
  isimler = rehber.keys()
  giris = input("Kişi adı : ")
  giris = giris.lower()

  if  giris in isimler:
    tel = input("Hangi telefon numarasını istiyorsunuz : ")
    tel = tel.lower()
    print(rehber.get(giris).get(tel, "İstediğiniz numara mevcut değildir."))
  else:
    print("İstediğiniz kişi mevcut değildir.")
  cikis = input("Yeniden arama yapmak için enter'a basınız veya çıkış için Q tuşuna basınız.")
  if cikis == "Q" or cikis == "q":
    print("Çıkış yapılıyor.")
    break
```
